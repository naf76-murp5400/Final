# Mixed-Use Development Strategy  
### Capital Allocation & Public-Private Investment Recommendation  

This notebook evaluates three mixed-use development opportunities:

- Waterfront DC (Urban Infill)  
- Landover Mall (Redevelopment)  
- South Riding, VA (Greenfield Growth)  

The objective is to identify a capital allocation strategy that balances:
- Financial return  
- Risk exposure  
- Public-private partnership potential  


```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Site": ["Waterfront DC", "Landover Mall", "South Riding"],
    "Total Cost": [79695219, 363815909, 396889807],
    "NOI": [4916756.36, 16102394.93, 21698617.02],
    "RLV": [17996800, 26525312, 141637683]
}

df = pd.DataFrame(data)
df
```

## Financial Comparison

The table below summarizes total development cost, net operating income (NOI), and residual land value (RLV).

```python
plt.figure()
plt.bar(df["Site"], df["NOI"])
plt.title("Net Operating Income Comparison")
plt.ylabel("NOI ($)")
plt.xticks(rotation=20)
plt.show()
```

```python
plt.figure()
plt.bar(df["Site"], df["Total Cost"])
plt.title("Total Development Cost Comparison")
plt.ylabel("Cost ($)")
plt.xticks(rotation=20)
plt.show()
```

```python
plt.figure()
plt.bar(df["Site"], df["RLV"])
plt.title("Residual Land Value Comparison")
plt.ylabel("RLV ($)")
plt.xticks(rotation=20)
plt.show()
```

```python
import numpy as np

rent_change = np.linspace(-0.2, 0.2, 10)
noi_base = df["NOI"]

plt.figure()
for i, site in enumerate(df["Site"]):
    noi_sensitivity = noi_base[i] * (1 + rent_change)
    plt.plot(rent_change, noi_sensitivity, label=site)

plt.title("NOI Sensitivity to Rent Changes")
plt.xlabel("Rent Change (%)")
plt.ylabel("NOI ($)")
plt.legend()
plt.show()
```

```python
cost_change = np.linspace(-0.2, 0.2, 10)
rlv_base = df["RLV"]

plt.figure()
for i, site in enumerate(df["Site"]):
    rlv_sensitivity = rlv_base[i] * (1 - cost_change)
    plt.plot(cost_change, rlv_sensitivity, label=site)

plt.title("RLV Sensitivity to Cost Changes")
plt.xlabel("Cost Change (%)")
plt.ylabel("Residual Land Value ($)")
plt.legend()
plt.show()
```

## Development Strategy

### South Riding, VA — Flagship Investment  
- Highest NOI and RLV  
- Strong residential-driven retail demand  
- Primary return driver  

### Waterfront DC — Strategic Urban Investment  
- Lower cost, stable returns  
- Aligns with smart growth and transit-oriented development  

### Landover Mall — Public-Private Opportunity  
- High cost and risk  
- Requires incentives (TIF, grants, partnerships)  


## Final Recommendation

Approve a phased capital allocation strategy:

- **1 → South Riding** (Return Engine)  
- **2 → Waterfront DC** (Stability + Urban Impact)  
- **3% → Landover Mall** (Conditional on public incentives)  

This approach maximizes return while balancing risk and public impact.
